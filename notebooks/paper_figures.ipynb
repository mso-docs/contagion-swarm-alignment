```python
from pathlib import Path
import json, sys, os
import pandas as pd
from IPython.display import display, Image
repo = Path.cwd() if Path("report").exists() else Path.cwd().parent
sys.path.insert(0, str(repo / "scripts"))
from paper_analysis import figures, sha
evidence = repo / "report/data/compact_evidence.json"
output = Path(os.environ.get("CONTAGION_PAPER_OUT", str(repo / "results/paper-draft"))) / "figures"
data = json.loads(evidence.read_text())
runs = pd.DataFrame(data["runs"])
display(runs.groupby(["family", "condition", "status"]).size().rename("attempts").to_frame())
print("Frozen evidence SHA256:", sha(evidence))
print("Complete runs:", int(runs.status.eq("completed").sum()), "all attempts:", len(runs))
manifest = figures(evidence, output)
```
